# Read in MNIST dataset (all 60,000 images from training and 10,000 from test).  Use scikitlearn to reduce with Principal Component Analysis from 784 dimensions to three.  Then use plotly to make an interactive plot.

```python
from tensorflow.keras.datasets import mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
# import PCA library
from sklearn.decomposition import PCA  
import numpy as np

# Reshape the data to 2D
train_images = train_images.reshape((60000, 28*28))
test_images = test_images.reshape((10000, 28*28))  

# Normalize the data to [0, 1]
train_images = train_images.astype('float32') / 255 
test_images = test_images.astype('float32') / 255

# use PCA to reduce dimensions to 3D
pca = PCA(n_components=3)
pca.fit(train_images)
train_images_pca = pca.transform(train_images) 
test_images_pca = pca.transform(test_images)

# use plotly to visualize the data
import plotly.express as px
import pandas as pd
import plotly.io as pio
pio.renderers.default = "browser"
df = pd.DataFrame(data=train_images_pca, columns=['x', 'y', 'z'])
df['label'] = train_labels
fig = px.scatter_3d(df, x='x', y='y', z='z', color='label', title="PCA of MNIST train Dataset")
fig.update_traces(marker=dict(size=2))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()

```

```python
# bigger markers
df = pd.DataFrame(data=train_images_pca, columns=['x', 'y', 'z'])
df['label'] = train_labels
fig = px.scatter_3d(df, x='x', y='y', z='z', color='label', title="PCA of MNIST train Dataset")
fig.update_traces(marker=dict(size=4))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()

df = pd.DataFrame(data=test_images_pca, columns=['x', 'y', 'z'])
df['label'] = test_labels
fig = px.scatter_3d(df, x='x', y='y', z='z', color='label', title="PCA of MNIST test Dataset")
fig.update_traces(marker=dict(size=4))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()
```

# Load fashion MNSI dataset and perform PCA

    Fashion MNSIT in 3D:
    
    0: T-shirt/top
    1: Trouser
    2: Pullover
    3: Dress
    4: Coat
    5: Sandal
    6: Shirt
    7: Sneaker
    8: Bag
    9: Ankle boot

```python
# again using fashion MNIST
import keras
(train_images, train_labels), (test_images, test_labels) = keras.datasets.fashion_mnist.load_data()
# Reshape the data to 2D
train_images = train_images.reshape((60000, 28*28))
test_images = test_images.reshape((10000, 28*28))  

# Normalize the data to [0, 1]
train_images = train_images.astype('float32') / 255 
test_images = test_images.astype('float32') / 255

# use PCA to reduce dimensions to 3D
pca = PCA(n_components=3)
pca.fit(train_images)
train_images_pca = pca.transform(train_images) 
test_images_pca = pca.transform(test_images)

df = pd.DataFrame(data=train_images_pca, columns=['x', 'y', 'z'])
df['label'] = train_labels
fig = px.scatter_3d(df, x='x', y='y', z='z', color='label', title="PCA of fashion MNIST train Dataset")
fig.update_traces(marker=dict(size=2))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()
```

```python
# bigger markers
fig = px.scatter_3d(df, x='x', y='y', z='z', color='label', title="PCA of fashion MNIST train Dataset")
fig.update_traces(marker=dict(size=4))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()
```

# Just for fun, used PCA to reduce to 4 dimensions and then used plotly to make a time animation of the data with the 4th PCA dimension being time.  The time value is the integer bin number, not the actual value since plotly works better with integers for time steps.

```python
# use PCA to reduce dimensions to 4D
pca = PCA(n_components=4)
pca.fit(train_images)
train_images_pca = pca.transform(train_images) 
test_images_pca = pca.transform(test_images)

df = pd.DataFrame(data=train_images_pca, columns=['x', 'y', 'z', 't'])
df['label'] = train_labels
x = pd.cut(df['t'],20, retbins=True, labels = False)
print(x[0])
df['t'] = x[0]
df = df.sort_values(by='t')
fig = px.scatter_3d(df, x='x', y='y', z='z', animation_frame="t", animation_group = 'label', color='label', title="PCA of fashion MNIST train Dataset")
fig.update_traces(marker=dict(size=2))
fig.update_layout(scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='z'))    
fig.show()
```
